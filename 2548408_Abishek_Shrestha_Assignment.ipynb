Task 6 : Adding HDI Category Column:

```python
def hdi_category(hdi):
    if hdi < 0.550:
        return "Low"
    elif hdi < 0.700:
        return "Medium"
    elif hdi < 0.800:
        return "High"
    else:
        return "Very High"

hdi_2022_df['HDI Category'] = hdi_2022_df['hdi'].apply(hdi_category)
```

```python
hdi_2022_df["HDI_Category"] = hdi_2022_df["hdi"].apply(hdi_category)
```

Problem- 1B- HDI Visualization and Trend Analysis (2020– 2022)

Task 1 : Data Extraction and Saving:

```python
hdi_1b = df[df['year'].isin([2020, 2021, 2022])]
hdi_1b.to_csv("HDI_problem1B.csv", index=False)
```

Task 2 : Data Cleaning:

```python
hdi_1b = hdi_1b.replace("–", np.nan)
hdi_1b = hdi_1b.dropna(subset=['hdi', 'country', 'year'])
hdi_1b['hdi'] = pd.to_numeric(hdi_1b['hdi'], errors='coerce')
hdi_1b = hdi_1b.drop_duplicates()
```

Task 3 :

  A : Visualization Tasks:

```python
countries = ['Nepal', 'India', 'China', 'United States', 'Germany']
data_line = hdi_1b[hdi_1b['country'].isin(countries)]

plt.figure(figsize=(8,5))
sns.lineplot(data=data_line, x='year', y='hdi', hue='country', marker='o')
plt.title("HDI Trend (2020–2022)")
plt.xlabel("Year")
plt.ylabel("HDI")
plt.show()
```

 B. Generate Visualizations:

– Bar Chart: Average HDI by Region (2020–2022)

```python
south_asia = ["Afghanistan", "Bangladesh", "Bhutan", "India",
              "Maldives", "Nepal", "Pakistan", "Sri Lanka"]
middle_east = ["Bahrain", "Iran", "Iraq", "Israel", "Jordan",
               "Kuwait", "Lebanon", "Oman", "Palestine",
               "Qatar", "Saudi Arabia", "Syria",
               "United Arab Emirates", "Yemen"]

def assign_region(country):
    if country in south_asia:
        return 'South Asia'
    elif country in middle_east:
        return 'Middle East'
    else:
        return 'Other' # Assign 'Other' for countries not in these specific lists

hdi_1b['region'] = hdi_1b['country'].apply(assign_region)

region_avg = hdi_1b.groupby(['region','year'])['hdi'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(data=region_avg, x='region', y='hdi', hue='year')
plt.xticks(rotation=45)
plt.title("Average HDI by Region (2020–2022)")
plt.show()
```

– Box Plot: HDI Distribution for 2020, 2021, and 2022



```python
plt.figure(figsize=(6,5))
sns.boxplot(data=hdi_1b, x='year', y='hdi')
plt.title("HDI Distribution (2020–2022)")
plt.show()
```

– Scatter Plot: HDI vs. GNI per Capita

```python
if 'gross inc percap' in hdi_1b.columns:
    plt.figure(figsize=(6,5))
    sns.scatterplot(data=hdi_1b, x='gross inc percap', y='hdi')
    plt.title("HDI vs GNI per Capita")
    plt.show()
else:
    print("GNI per Capita variable not available in the dataset.")
```

 Problem 2

Advanced HDI Exploration

Task 1 :Create South Asia Subset:

```python
south_asia = ["Afghanistan", "Bangladesh", "Bhutan", "India",
              "Maldives", "Nepal", "Pakistan", "Sri Lanka"]

sa_df = df[df['country'].isin(south_asia)].copy()
sa_df.to_csv("HDI_SouthAsia.csv", index=False)
```

Task 2 :  Composite Development Score:

```python
sa_df['Composite Score'] = (
    0.30 * sa_df['life_expectancy'] +
    0.30 * sa_df['gross_inc_percap']
)

sa_df.sort_values(by='Composite Score', ascending=False).head()
```

```python
# STEP 1: Create Composite Score
sa_df['Composite Score'] = (
    0.30 * sa_df['life_expectancy'] +
    0.30 * sa_df['gross_inc_percap']
)

# STEP 2: Create Composite Rank (higher score = better rank)
sa_df['Composite Rank'] = sa_df['Composite Score'].rank(
    ascending=False,
    method='dense'
)

# STEP 3: Create HDI Rank (higher HDI = better rank)
sa_df['HDI Rank'] = sa_df['hdi'].rank(
    ascending=False,
    method='dense'
)

# STEP 4: Create final merged ranking table
ranking_table = sa_df[
    ['country', 'Composite Score', 'Composite Rank', 'hdi', 'HDI Rank']
].sort_values(by='Composite Rank')

# STEP 5: Display top 5 countries
print("Top 5 South Asian Countries: Composite Rank vs HDI Rank\n")
ranking_table.head(5)
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=sa_df.sort_values('Composite Score', ascending=False).head(5),
            x='Composite Score', y='country')
plt.title("Top 5 South Asian Countries by Composite Score")
plt.show()
```

Task 3 : Outlier Detection:

```python
Q1 = sa_df['hdi'].quantile(0.25)
Q3 = sa_df['hdi'].quantile(0.75)
IQR = Q3 - Q1

outliers = sa_df[(sa_df['hdi'] < Q1 - 1.5*IQR) | (sa_df['hdi'] > Q3 + 1.5*IQR)]
```

```python
plt.figure(figsize=(6,5))
sns.scatterplot(data=sa_df, x='gross_inc_percap', y='hdi')
sns.scatterplot(data=outliers, x='gross_inc_percap', y='hdi', color='red')
plt.title("Outliers in South Asia")
plt.show()
```

Task 4 :  Exploring Metric Relationships:

```python
correlation_matrix = sa_df[['hdi', 'gender_development', 'life_expectancy']].corr()
correlation_matrix
```

```python
plt.figure(figsize=(6,5))
sns.regplot(data=sa_df, x='life_expectancy', y='hdi')
plt.title("HDI vs Life Expectancy (South Asia)")
plt.xlabel("Life Expectancy Index")
plt.ylabel("HDI")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.regplot(data=sa_df, x='gender_development', y='hdi')
plt.title("HDI vs Gender Development Index (South Asia)")
plt.xlabel("Gender Development Index")
plt.ylabel("HDI")
plt.show()
```

Task 5 : Gap Analysis:

```python
# Create GNI–HDI Gap column
sa_df['GNI_HDI_Gap'] = sa_df['gross_inc_percap'] - sa_df['hdi']

# View first few rows
sa_df[['country', 'gross_inc_percap', 'hdi', 'GNI_HDI_Gap']].head()
```

```python
gap_desc = sa_df.sort_values(by='GNI_HDI_Gap', ascending=False)
gap_desc[['country', 'GNI_HDI_Gap']].head()
```

```python
top_positive_gap = gap_desc.head(3)
gap_asc = sa_df.sort_values(by='GNI_HDI_Gap', ascending=True)
top_negative_gap = gap_asc.head(3)
```

```python
plt.figure(figsize=(6,4))
sns.barplot(
    data=top_positive_gap,
    x='GNI_HDI_Gap',
    y='country'
)
plt.title("Top 3 Positive GNI–HDI Gaps (South Asia)")
plt.xlabel("GNI–HDI Gap")
plt.ylabel("Country")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(
    data=top_negative_gap,
    x='GNI_HDI_Gap',
    y='country'
)
plt.title("Top 3 Negative GNI–HDI Gaps (South Asia)")
plt.xlabel("GNI–HDI Gap")
plt.ylabel("Country")
plt.show()
```
